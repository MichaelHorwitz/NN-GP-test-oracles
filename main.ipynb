```python
from collections import defaultdict
from matplotlib import pyplot as plt
from software_under_test.Collatz import Collatz
import seaborn as sns

input_set = [x for x in range(4000)]
sut: Collatz = Collatz()
output_set = [sut.exec_real_output(x) for x in input_set]
min_val = min(output_set)
max_val = max(output_set)
bins = range(min_val // 10 * 10, (max_val + 10) // 10 * 10 + 10, 10)
plt.hist(output_set, bins=bins, edgecolor='black')
plt.xlabel("Output")
plt.show()
```

```python
plt.figure(figsize=(6.85,3.82))
plt.scatter(input_set, output_set, s=1)
plt.xlabel("Input value")
plt.ylabel("Output value")
plt.title("Distribution of outputs for modified Collatz inputs")
plt.show()
```

```python
from collections import defaultdict
import random
binned_data = defaultdict(list)
for output in output_set:
    if output >= 60:
        continue
    key = (output // 10) * 10
    binned_data[key].append(output)
for key in sorted(binned_data.keys()):
    print(f"  Bin {key}: {binned_data[key]}")
fair_list = [random.sample(binned_data[key], 5) for key in binned_data.keys()]
fair_list = [item for sublist in fair_list for item in sublist]
out_in_dict = {}
for inp, out in zip(input_set, output_set):
    out_in_dict[out] = inp
print(fair_list)
fair_inp = [out_in_dict[out] for out in fair_list]
print(fair_inp)
```
